```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression

model = LinearRegression()
```

```python
X = np.array([[1], [2], [3], [4], [5]])

y = np.array([35, 42, 50, 58, 65])
```

```python
X.shape
```

```python
y.shape
```

```python
model.fit( X, y )
```

```python
predictions = model.predict(X)
```

```python
slope = model.coef_[0]

Intercept = model.intercept_
```

```python
print("Slope: ", slope)
print("Intercept: ", Intercept)
print("Predictions: ",predictions)
```

```python
residual = y - predictions

print("Residual: ", residual)
```

```python
mae = np.absolute(residual).mean()

print(mae)
```

```python
mse = (residual ** 2).mean()

print(mse)
```

```python
rsme = mse ** 0.5

print(rsme)
```

```python
plt.scatter( X, y)
```

```python
plt.scatter( X, predictions)

plt.plot( X, predictions)

plt.xlabel( "Hours Studied")

plt.ylabel( "Marks Get")

plt.show()
```

## 6. Gradient Descent : 
* It calculates the partial derivatives of the loss functin wrt the model parameters, and uses them to update the parameters in the direction that decreases the loss. 
* Gradient means when we find the derivative of single variable (sum of Squares) wrt two variables (intercept and slope). Descent means to minimise the Loss function.  (not actual but mine)
