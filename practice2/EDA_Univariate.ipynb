#### sex
##### Conclusion
* 2 categories : Male and Female
* Males are in majority

```python
tip["sex"].value_counts()
```

```python
sns.catplot(data=tip, x="sex", kind="count")
```

```python
x=tip["sex"].value_counts()
plt.pie(x, labels=["Male","Female"], autopct="%1.1f%%")
plt.show()
```

#### Smoker
